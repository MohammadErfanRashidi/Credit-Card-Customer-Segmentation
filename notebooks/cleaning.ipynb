```python
# Import the libraries 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv('../data/credit_card.csv')

# Display the first few rows
df.head()
```

```python
# Info
df.info()
```

```python
# Describe 
df.describe()
```

```python
# Check for null values
df.isnull().sum()
```

```python
df.dropna(inplace = True)

# Check null values again
df.isnull().sum().sum()
```

```python
# Strip and lowercase columns for simplicity 
df.columns = df.columns.str.strip().str.lower()

df.columns.tolist()
```

```python
# Check the datatypes
df.dtypes
```

```python
# Out of range check
freq_cols = ['balance_frequency', 'purchases_frequency', 'oneoff_purchases_frequency',
             'purchases_installments_frequency', 'cash_advance_frequency']

for col in freq_cols:
    out = df[(df[col] > 1) | (df[col] < 0)]
    print(f'{col}: {len(out)} rows outside [0, 1]')
    if len(out) > 0:
        display(out[['cust_id', col]])
```

We are going to replace these values above 1 with 1 in order to fix the issue.

```python
# Cap above 1 values to 1
df['cash_advance_frequency'] = df['cash_advance_frequency'].clip(upper = 1, lower = 0)

# Check the first instance in the table above, it must be 1 now
df[df['cust_id'] == 'C10708']['cash_advance_frequency']
```

```python
# Percentage full payment 
out_prc = df[(df['prc_full_payment'] > 1) | (df['prc_full_payment'] < 0)]
print(f'prc: {len(out_prc)} rows outside [0, 1]')
if len(out_prc) > 0:
    display(out_prc[['cust_id', 'prc_full_payment']])
```

```python
# Check for negative values in monetary columns
money_cols = ['balance', 'purchases', 'oneoff_purchases', 'installments_purchases',
              'cash_advance', 'payments', 'minimum_payments', 'credit_limit']

for col in money_cols:
    neg_count = (df[col] < 0).sum()
    print(f'{col}: {neg_count} negative values')
```

```python
# Credit card
zero_limit = (df['credit_limit'] == 0).sum()
neg_limit = (df['credit_limit'] < 0).sum()

# Tenure limits
zero_ten = (df['tenure'] == 0).sum()
neg_ten = (df['tenure'] < 0).sum()

print(f'credit: {zero_limit} 0s, and {neg_limit} negative values.')
print(f'tenure: {zero_ten} 0s, and {neg_ten} negative values.')
```

```python
# Drop cust_id column 
df.drop(columns = ['cust_id'], inplace = True)
df.info()
```

```python
# Plot outliers with boxplot
fig, axes = plt.subplots(nrows = 6, ncols = 3, figsize = (12, 18))
axes = axes.flatten()

for i, col in enumerate(df.columns):
    data = df[col]
    axes[i].boxplot(data, vert = True, patch_artist = True,
                    boxprops = dict(facecolor = 'lightblue', color = 'black'),
                    flierprops = dict(marker = 'o', markerfacecolor = 'red', markersize = 3))
    axes[i].set_title(col)
    axes[i].set_ylabel('Value')

plt.tight_layout()
plt.show()
```

We are going to keep outliers as they are and investigate further in EDA notebook.

```python
# Logical consistency for purchase 
df['purchase_diff'] = df['purchases'] - (df['oneoff_purchases'] + df['installments_purchases'])
tolerance = 0.01
inconsistent_purchases = df[abs(df['purchase_diff']) > tolerance]
print(f'Purchase mismatch: {len(inconsistent_purchases)}')

valid_payment = df.dropna(subset = ['payments', 'minimum_payments'])
inconsistent_payments = valid_payment[valid_payment['payments'] < valid_payment['minimum_payments']]
print(f'Where payments less than minimum payments: {len(inconsistent_payments)}')
```

Before we decide for the fix, we first examine the inconsistencies.

```python
if len(inconsistent_purchases) > 0:
    display(inconsistent_purchases[['purchases', 'oneoff_purchases', 'installments_purchases', 'purchase_diff']])
```

```python
if len(inconsistent_payments) > 0:
    display(inconsistent_payments[['payments', 'minimum_payments']])
    zero_payments = (inconsistent_payments['payments'] == 0).sum()
    print(f'Rows with payments equal to 0 are {zero_payments}')
```

We are going to drop these contradictory rows.

```python
drop_indices = pd.Index(inconsistent_purchases.index).union(inconsistent_payments.index)

df = df.drop(index = drop_indices)

# Remove the helper column
df.drop(columns = ['purchase_diff'], inplace = True)

# New shape of the dataset
print(df.shape)

# First five rows
df.head()
```

```python
# With purchase more than 0, purchase freq must also be more than 0
print(len(df[(df['purchases'] > 0) & (df['purchases_frequency'] == 0)]))
print(len(df[(df['purchases'] == 0) & (df['purchases_frequency'] != 0)]))
```

```python
# Same logic with cash
print(len(df[(df['cash_advance'] > 0) & (df['cash_advance_frequency'] == 0)]))
print(len(df[(df['cash_advance'] == 0) & (df['cash_advance_frequency'] != 0)]))
```

```python
# Same logic with oneoff
print(len(df[(df['oneoff_purchases'] > 0) & (df['oneoff_purchases_frequency'] == 0)]))
print(len(df[(df['oneoff_purchases'] == 0) & (df['oneoff_purchases_frequency'] != 0)]))
```

```python
# Same logic with installment
print(len(df[(df['installments_purchases'] > 0) & (df['purchases_installments_frequency'] == 0)]))
print(len(df[(df['installments_purchases'] == 0) & (df['purchases_installments_frequency'] != 0)]))
```

We covered all the logical issue that could be in the dataset.
